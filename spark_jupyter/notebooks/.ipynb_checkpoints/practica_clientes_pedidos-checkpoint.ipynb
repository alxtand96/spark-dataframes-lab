# Práctica · Spark DataFrames · Clientes y pedidos

Completa las celdas indicadas. Trabaja sobre el entorno dockerizado incluido en `spark_jupyter/`.

```python

from iniciar_spark import get_spark
from pyspark.sql import functions as F

spark = get_spark("PracticaClientesPedidos")
print("SparkSession creada")
print("Versión de Spark:", spark.version)
```

```python
from pathlib import Path

data_dir = Path("/opt/spark-apps/datos")
clientes_path = data_dir / "clientes.csv"
pedidos_path = data_dir / "pedidos.csv"

print("Ruta clientes:", clientes_path)
print("Ruta pedidos:", pedidos_path)
```

## 1. Lee ambos CSV y muestra su esquema y algunas filas.

```python
# ============================================================
# TAREA 1 - LECTURA DE DATOS
# ============================================================

# Lectura del fichero de clientes
# - header=True: la primera fila contiene los nombres de columna
# - inferSchema=True: Spark detecta automáticamente el tipo de dato
# - sep=";": separador de campos punto y coma
df_clientes = spark.read.csv(
    str(clientes_path),
    header=True,
    inferSchema=True,
    sep=";"
)

# Lectura del fichero de pedidos
df_pedidos = spark.read.csv(
    str(pedidos_path),
    header=True,
    inferSchema=True,
    sep=";"
)

print(f"✅ clientes.csv cargado  → {df_clientes.count()} registros")
print(f"✅ pedidos.csv  cargado  → {df_pedidos.count()} registros")
```

```python
print("=" * 55)
print("ESQUEMA DE CLIENTES")
print("=" * 55)
df_clientes.printSchema()
df_clientes.show(5, truncate=False)

print("=" * 55)
print("ESQUEMA DE PEDIDOS")
print("=" * 55)
df_pedidos.printSchema()
df_pedidos.show(5, truncate=False)
```

## 2. Limpia el DataFrame de clientes: trim en ciudad y dropDuplicates.

```python
# ============================================================
# TAREA 2 - LIMPIEZA DE DATOS
# ============================================================

# --- 2.1 TRIM: eliminar espacios en columnas de texto ---
# Detectamos que ciudad tiene espacios extra al inicio y final
# También aplicamos trim a nombre y segmento por consistencia
df_clientes = df_clientes.withColumn("ciudad",   F.trim(F.col("ciudad"))) \
                         .withColumn("nombre",   F.trim(F.col("nombre"))) \
                         .withColumn("segmento", F.trim(F.col("segmento")))

print("✅ Trim aplicado en columnas de texto")
print("\nCiudades tras trim:")
df_clientes.show(5, truncate=False)

# --- 2.2 DUPLICADOS ---
total_antes = df_clientes.count()
df_clientes = df_clientes.dropDuplicates()
total_despues = df_clientes.count()

print(f"\n✅ Duplicados eliminados: {total_antes - total_despues}")
print(f"   Registros antes:  {total_antes}")
print(f"   Registros después: {total_despues}")

# --- 2.3 VALORES NULOS ---
print("\n✅ Valores nulos por columna (clientes):")
df_clientes.select(
    [F.count(F.when(F.col(c).isNull(), c)).alias(c) for c in df_clientes.columns]
).show()

print("✅ Valores nulos por columna (pedidos):")
df_pedidos.select(
    [F.count(F.when(F.col(c).isNull(), c)).alias(c) for c in df_pedidos.columns]
).show()

# Eliminamos filas con nulos en campos clave
df_clientes = df_clientes.dropna(subset=["id_cliente", "nombre"])
df_pedidos  = df_pedidos.dropna(subset=["id_pedido", "id_cliente"])

print("\n✅ Limpieza completada")
print(f"   Clientes limpios: {df_clientes.count()} registros")
print(f"   Pedidos limpios:  {df_pedidos.count()} registros")
```

```python
# --- 2.4 TRATAMIENTO DE NULOS EN PEDIDOS ---

# Rellenamos producto nulo con "Desconocido"
# Rellenamos cantidad nula con 0.0
df_pedidos = df_pedidos.fillna({"producto": "Desconocido", "cantidad": 0.0})

print("✅ Nulos en pedidos tratados:")
df_pedidos.select(
    [F.count(F.when(F.col(c).isNull(), c)).alias(c) for c in df_pedidos.columns]
).show()

print(f"Pedidos tras limpieza completa: {df_pedidos.count()} registros")
```

## 3. Convierte tipos en pedidos y crea la columna `importe`.

```python
# ============================================================
# TAREA 3 - TRANSFORMACIÓN
# ============================================================

# --- 3.1 CONVERSIÓN DE TIPOS ---
# cantidad viene como Double, lo convertimos a Integer para consistencia
# precio_unitario ya es Integer, lo casteamos a Double para el cálculo
df_pedidos = df_pedidos.withColumn("cantidad",        F.col("cantidad").cast("integer")) \
                       .withColumn("precio_unitario", F.col("precio_unitario").cast("double"))

# --- 3.2 NUEVA COLUMNA: importe ---
# importe = cantidad × precio_unitario
# Representa el valor económico total de cada línea de pedido
df_pedidos = df_pedidos.withColumn(
    "importe",
    F.round(F.col("cantidad") * F.col("precio_unitario"), 2)
)

print("✅ Tipos convertidos y columna importe creada")
print("\nEsquema actualizado de pedidos:")
df_pedidos.printSchema()

print("\nMuestra con columna importe:")
df_pedidos.select("id_pedido", "id_cliente", "producto",
                  "cantidad", "precio_unitario", "importe").show(8, truncate=False)

print(f"\nImporte máximo:  {df_pedidos.agg(F.max('importe')).collect()[0][0]}")
print(f"Importe mínimo:  {df_pedidos.agg(F.min('importe')).collect()[0][0]}")
print(f"Importe medio:   {round(df_pedidos.agg(F.avg('importe')).collect()[0][0], 2)}")
```

## 4. Haz un join entre clientes y pedidos.

```python
# ============================================================
# TAREA 4 - INTEGRACIÓN DE DATOS (JOIN)
# ============================================================

# Usamos inner join: solo conserva registros con correspondencia
# en ambas tablas (id_cliente presente en clientes Y en pedidos)
df_join = df_clientes.join(df_pedidos, on="id_cliente", how="inner")

print("✅ JOIN realizado correctamente")
print(f"\nRegistros tras inner join: {df_join.count()}")
print(f"Clientes originales:       {df_clientes.count()}")
print(f"Pedidos originales:        {df_pedidos.count()}")

print("\nMuestra del resultado del join:")
df_join.select("id_cliente", "nombre", "ciudad", "segmento",
               "id_pedido", "producto", "cantidad", "importe").show(8, truncate=False)

# --- ANÁLISIS DE REGISTROS PERDIDOS ---
# Hacemos un left join para detectar clientes sin pedidos
df_left = df_clientes.join(df_pedidos, on="id_cliente", how="left")
sin_pedidos = df_left.filter(F.col("id_pedido").isNull())

print(f"\n⚠️  Clientes sin pedidos (perdidos en inner join): {sin_pedidos.count()}")
sin_pedidos.select("id_cliente", "nombre", "ciudad", "segmento").show(truncate=False)
```

## 5. Filtra ventas Premium con importe >= 100.

```python
# ============================================================
# TAREA 5 - FILTRAR VENTAS PREMIUM CON IMPORTE >= 100
# ============================================================

df_premium = df_join.filter(
    (F.col("segmento") == "Premium") & (F.col("importe") >= 100)
)

print(f"✅ Pedidos Premium con importe >= 100: {df_premium.count()}")
df_premium.select("nombre", "ciudad", "segmento",
                  "producto", "importe").show(5, truncate=False)
```

## 6. Clasifica pedidos con `when` en Alto / Medio / Bajo.

```python

# ============================================================
# TAREA 6 - CLASIFICAR PEDIDOS CON when: Alto / Medio / Bajo
# ============================================================

# Clasificamos según el importe total del pedido:
# Alto  → importe >= 500
# Medio → importe entre 100 y 499
# Bajo  → importe < 100
df_join = df_join.withColumn(
    "categoria_importe",
    F.when(F.col("importe") >= 500, "Alto")
     .when(F.col("importe") >= 100, "Medio")
     .otherwise("Bajo")
)

print("\n✅ Clasificación de pedidos por importe:")
df_join.groupBy("categoria_importe") \
       .count() \
       .orderBy("categoria_importe") \
       .show()

print("\nMuestra con clasificación:")
df_join.select("nombre", "producto", "importe",
               "categoria_importe").show(8, truncate=False)
```

## 7. Calcula agregaciones por ciudad y segmento.

```python
# ============================================================
# TAREA 7 - ANÁLISIS Y AGREGACIONES
# ============================================================

# Agrupamos por ciudad y segmento para obtener métricas de negocio
df_agregado = df_join.groupBy("ciudad", "segmento") \
    .agg(
        F.count("id_pedido").alias("num_pedidos"),
        F.round(F.sum("importe"), 2).alias("ingresos_totales"),
        F.round(F.avg("importe"), 2).alias("importe_medio"),
        F.round(F.max("importe"), 2).alias("importe_maximo")
    ) \
    .orderBy("ingresos_totales", ascending=False)

print("✅ Agregaciones por ciudad y segmento:")
df_agregado.show(20, truncate=False)

# Resumen global por segmento
print("=" * 50)
print("RESUMEN GLOBAL POR SEGMENTO:")
df_join.groupBy("segmento") \
    .agg(
        F.count("id_pedido").alias("num_pedidos"),
        F.round(F.sum("importe"), 2).alias("ingresos_totales"),
        F.round(F.avg("importe"), 2).alias("importe_medio")
    ) \
    .orderBy("ingresos_totales", ascending=False) \
    .show()
```

## 8. Crea una vista temporal y haz una consulta SQL.

```python
# ============================================================
# TAREA 8 - SQL SOBRE DATAFRAMES
# ============================================================

# Creamos una vista temporal para poder consultar con SQL estándar
# La vista existe solo durante la sesión de Spark activa
df_join.createOrReplaceTempView("ventas")

print("✅ Vista temporal 'ventas' creada")

# Consulta SQL: top 5 clientes por ingresos totales
# con número de pedidos y ticket medio
resultado_sql = spark.sql("""
    SELECT
        nombre,
        ciudad,
        segmento,
        COUNT(id_pedido)      AS num_pedidos,
        ROUND(SUM(importe), 2) AS ingresos_totales,
        ROUND(AVG(importe), 2) AS ticket_medio
    FROM ventas
    GROUP BY nombre, ciudad, segmento
    ORDER BY ingresos_totales DESC
    LIMIT 10
""")

print("\n✅ Top 10 clientes por ingresos totales:")
resultado_sql.show(truncate=False)
```

## 9. Usa `sample()` y `randomSplit()`.

```python
# ============================================================
# TAREA 9 - MUESTREO Y PARTICIÓN
# ============================================================

# --- 9.1 sample() ---
# Extrae una muestra aleatoria del DataFrame sin reemplazo
# fraction=0.3 → aproximadamente el 30% de los registros
# seed=42 → semilla fija para reproducibilidad
muestra = df_join.sample(withReplacement=False, fraction=0.3, seed=42)

print(f"✅ sample() aplicado:")
print(f"   DataFrame original: {df_join.count()} registros")
print(f"   Muestra obtenida:   {muestra.count()} registros (≈30%)")
muestra.select("nombre", "ciudad", "producto", "importe").show(5, truncate=False)

# --- 9.2 randomSplit() ---
# Divide el DataFrame en dos subconjuntos: 80% entrenamiento, 20% test
# Útil para preparar datos para modelos de Machine Learning
train, test = df_join.randomSplit([0.8, 0.2], seed=42)

print(f"\n✅ randomSplit() aplicado (80% / 20%):")
print(f"   Conjunto train: {train.count()} registros")
print(f"   Conjunto test:  {test.count()} registros")
print(f"   Total:          {train.count() + test.count()} registros")
```

## 10. Guarda el resultado en Parquet en `/opt/spark-apps/salida/resultado_parquet` y léelo de nuevo.

```python
# ============================================================
# TAREA 10 - PERSISTENCIA EN FORMATO PARQUET
# ============================================================

parquet_path = "/opt/spark-apps/salida/resultado_parquet"

# --- 10.1 ESCRITURA ---
# Guardamos el DataFrame del join completo en formato Parquet
# mode="overwrite": sobreescribe si ya existe
# Parquet es un formato columnar binario, eficiente en compresión y lectura
df_join.write.mode("overwrite").parquet(parquet_path)

print(f"✅ Resultado guardado en Parquet:")
print(f"   Ruta: {parquet_path}")
print(f"   Registros escritos: {df_join.count()}")

# --- 10.2 LECTURA ---
# Leemos de nuevo el fichero Parquet para verificar la escritura
df_parquet = spark.read.parquet(parquet_path)

print(f"\n✅ Parquet leído correctamente:")
print(f"   Registros recuperados: {df_parquet.count()}")
print("\nEsquema del Parquet:")
df_parquet.printSchema()

print("\nMuestra del Parquet:")
df_parquet.select("nombre", "ciudad", "segmento",
                  "producto", "importe", "categoria_importe").show(5, truncate=False)
```

## 11. Responde brevemente en Markdown:
- ¿Qué ventaja tiene usar join?
        El join permite combinar información de dos fuentes de datos distintas 
        en un único DataFrame, enriqueciendo cada registro con datos de ambas 
        tablas. En este caso nos ha permitido asociar los datos del cliente 
        (nombre, ciudad, segmento) con cada uno de sus pedidos, habilitando 
        análisis que serían imposibles con las tablas por separado.

- ¿Qué diferencia hay entre sample y randomSplit?
        `sample()` extrae una fracción aleatoria del DataFrame original para 
        obtener una muestra representativa, manteniendo el DataFrame completo 
        intacto. `randomSplit()` en cambio divide el DataFrame en dos o más 
        subconjuntos mutuamente excluyentes que suman el total de registros, 
        lo que resulta útil para separar datos de entrenamiento y test en 
        procesos de Machine Learning.
- ¿Qué pedido se pierde en el inner join y por qué?
        En el inner join se pierden **10 pedidos** cuyos `id_cliente` no tienen 
        correspondencia en la tabla de clientes (por ejemplo pedidos de clientes 
        con id 34, 25 o 28 que fueron eliminados como duplicados o no existen). 
        Además se pierden **3 clientes** (Fernando, Nuria y Mario) que no tienen 
        ningún pedido asociado. El inner join solo conserva registros con 
        correspondencia en ambas tablas.
